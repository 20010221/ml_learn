```python
import random
import math
n = random.randint(1, 100)  # 函数 random.randint(a,b) 返回从 a 到 b 之间随机选出的一个整数
a, b, cnt = 1, 100, 0
while a < b :
    b = b-1
    m = math.ceil((a + b) / 2)  # 函数 math.ceil(x) 的返回值为数值 x 的向上取整
    if n < m:
        b = m - 1
    elif n > m :
        a = m
    else:
        cnt += 1
print(cnt)
```

```python
# 下面程序中的 Change 函数的功能是将十进制数转化 N 进制数（2 <= N <= 16）。
#  输入目标进制和待转换的数，函数返回转换后的目标进制数。
#  请在横线上填写代码并给出程序的输出结果。

def Change(radix, num):
    g = []
    while num != 0:
        r = num % radix
        num = num // radix
        g.append(r)
    ans = ''
    t = ('A', 'B', 'C', 'D', 'E', 'F')
    for x in range(len(g)):
        if g[x] >= 10:
            ans += t[x]
        else:
            ans += str(g[x])
    return
print(Change(16, 12455))
```

```python
# 4. 给定一个整数的数字列表 L，判断 L 中是否存在相同数字，若存在返回 True，否则返回 False。
#  请在横线上补全代码。
def identical(L):
    flag = 0
    L.sort()
    for i in range(len(L)):
        if L[i] == L[i+1]:
            flag = 1
            break
    if flag:
        return True
    else:
        return False
```

```python
import numpy as np
import tensorflow as tf
time_steps = 60
features = 1
samples = 200

# 构造输入数据 (samples, time_steps, features)
x_train = np.random.rand(samples, time_steps, features)  # 200个 60天股票价格
y_train = np.random.rand(samples, 1)                     # 200个 对应的第61天股票价格
x_test  = np.random.rand(10, time_steps, features)
# 构建5层Sequential神经网络预测股票价格，要求包含：
# ●2个LSTM层（首层units=4需设置input_shape）
# ●每个LSTM层后接Dropout层(rate=0.2)
# ●最终Dense层(units=1) 实现步骤：
# 1.使用stock_model变量构建模型
# 2.编译模型(loss=‘mean_squared_error’, optimizer=‘adam’)
# 3.训练配置(epochs=5, batch_size=16, verbose=0)
# 4.输出模型summary()
# 5.预测x_test并输出最后一列
# 注意：x_train, y_train, x_test不用再写代码改动
stock_model = tf.keras.models.Sequential([
    tf.keras.layers.LSTM(4,input_shape=(time_steps, features),return_sequences=True),
    tf.keras.layers.Dropout(rate = 0.2),
    tf.keras.layers.LSTM(samples,return_sequences=True),
    tf.keras.layers.Dropout(rate = 0.2),
    tf.keras.layers.Dense(1)
])
```

```python

stock_model.compile(optimizer='adam',loss="mean_squared_error",metrics=["accuracy"])
summary = stock_model.fit(x_train, y_train, epochs = 5, batch_size=16, verbose=0)
result = stock_model.predict(x_test[:-1])
```
